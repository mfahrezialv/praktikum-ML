# IDENTIFIKASI DATASET BERDASARKAN JENIS KLASIFIKASI

## A. Binary Classification (Klasifikasi Biner)

Pada klasifikasi biner, model Machine Learning bertujuan memprediksi data ke dalam dua kelas. Contohnya adalah memprediksi gagal bayar atau tidak gagal bayar.

### 1. Credit Risk Dataset

Kaggle: https://www.kaggle.com/datasets/laotse/credit-risk-dataset

- Tujuan: memprediksi apakah peminjam mengalami gagal bayar pinjaman atau tidak.
- Fitur (X): person_age, person_income, person_home_ownership, person_emp_length, loan_intent, loan_grade, loan_amnt, loan_int_rate, dan fitur lain yang tersedia.
- Target (y): loan_status.
- Label: 0 = tidak gagal bayar, 1 = gagal bayar.
- Identifikasi: termasuk Binary Classification karena target loan_status hanya mempunyai dua kelas, yaitu 0 dan 1.
- Alasan dipilih: dataset memiliki informasi keuangan dan pinjaman yang dapat digunakan untuk mempelajari cara kerja klasifikasi biner dalam memprediksi risiko kredit.

Catatan: dataset ini digunakan untuk pembelajaran Machine Learning, bukan sebagai satu-satunya dasar untuk menentukan kelayakan pinjaman seseorang.

### 2. Bank Marketing Subscription Prediction

Kaggle: https://www.kaggle.com/competitions/binary-classification-with-a-bank-database

- Tujuan: memprediksi apakah nasabah akan berlangganan deposito berjangka yang ditawarkan oleh bank.
- Fitur (X): informasi nasabah dan riwayat pemasaran yang tersedia pada dataset, seperti usia, pekerjaan, pendidikan, dan karakteristik kampanye.
- Target (y): y.
- Label: 0 = tidak berlangganan, 1 = berlangganan, sesuai pengodean target pada dataset kompetisi.
- Identifikasi: termasuk Binary Classification karena model hanya memprediksi dua kemungkinan hasil.
- Alasan dipilih: dataset ini menunjukkan penerapan klasifikasi biner dalam bidang pemasaran perbankan, yaitu memperkirakan keberhasilan penawaran produk kepada nasabah.

### 3. Credit Default Prediction Challenge

Kaggle: https://www.kaggle.com/competitions/credit-default-prediction-challenge-jcds-0612

- Tujuan: memprediksi apakah nasabah kartu kredit akan mengalami gagal bayar pada bulan berikutnya.
- Fitur (X): informasi nasabah, batas kredit, usia, pendidikan, status pembayaran sebelumnya, jumlah tagihan, dan jumlah pembayaran.
- Target (y): default_payment_next_month.
- Label: 0 = tidak gagal bayar pada bulan berikutnya, 1 = gagal bayar pada bulan berikutnya.
- Identifikasi: termasuk Binary Classification karena target hanya memiliki dua kemungkinan kelas, yaitu 0 dan 1.
- Alasan dipilih: dataset menyediakan informasi keuangan dan riwayat pembayaran yang dapat digunakan untuk mempelajari prediksi risiko gagal bayar.

## B. Multiclass Classification (Klasifikasi Multikelas)

Klasifikasi multikelas digunakan untuk memprediksi satu kelas dari tiga atau lebih kelas yang tersedia. Setiap sampel umumnya hanya mempunyai satu kelas target.

### 1. Forest Cover Type Prediction

Kaggle: https://www.kaggle.com/competitions/forest-cover-type-prediction

- Tujuan: memprediksi jenis tutupan hutan berdasarkan kondisi geografis dan karakteristik tanah.
- Fitur (X): Elevation, Aspect, Slope, jarak ke sumber air, jarak ke jalan, indeks hillshade, Wilderness_Area, dan Soil_Type.
- Target (y): Cover_Type.
- Label: terdapat tujuh kelas, yaitu Spruce/Fir, Lodgepole Pine, Ponderosa Pine, Cottonwood/Willow, Aspen, Douglas-fir, dan Krummholz.
- Identifikasi: termasuk Multiclass Classification karena target Cover_Type memiliki tujuh kelas yang berbeda.
- Alasan dipilih: dataset memiliki beberapa kelas target yang jelas sehingga cocok digunakan untuk mempelajari klasifikasi multikelas berdasarkan data geografis dan lingkungan.

### 2. Plant Seedlings Classification

Kaggle: https://www.kaggle.com/competitions/plant-seedlings-classification

- Tujuan: mengenali jenis tanaman berdasarkan gambar bibit tanaman.
- Fitur (X): gambar bibit tanaman yang memiliki karakteristik visual, seperti bentuk daun, warna, ukuran, dan pola pertumbuhan.
- Target (y): nama kelas atau jenis tanaman yang tercantum pada data pelatihan.
- Label: terdapat 12 jenis tanaman yang berbeda.
- Identifikasi: termasuk Multiclass Classification karena setiap gambar dikategorikan ke dalam satu jenis tanaman dari 12 kelas yang tersedia.
- Alasan dipilih: dataset ini cocok untuk mempelajari klasifikasi gambar dan memahami bagaimana model dapat membedakan beberapa jenis tanaman berdasarkan ciri visualnya.

### 3. Padang Food Image Classification

Kaggle: https://www.kaggle.com/competitions/coaching-1-padang-food-image-classification

- Tujuan: mengenali jenis makanan khas Padang berdasarkan gambar.
- Fitur (X): gambar makanan dalam format JPG atau PNG dengan karakteristik visual, seperti bentuk, warna, tekstur, dan tampilan makanan.
- Target (y): label atau nama kelas makanan pada data pelatihan.
- Label: terdapat sembilan kelas, yaitu ayam goreng, ayam pop, daging rendang, dendeng batokok, gulai ikan, gulai tambusu, gulai tunjang, telur balado, dan telur dadar.
- Identifikasi: termasuk Multiclass Classification karena model harus menentukan satu kelas makanan dari sembilan kelas yang tersedia.
- Alasan dipilih: dataset ini cocok untuk mempelajari klasifikasi gambar dengan contoh makanan Indonesia yang mudah dikenali dalam kehidupan sehari-hari.

## C. Multilabel Classification (Klasifikasi Multilabel)

Klasifikasi multilabel digunakan ketika satu sampel dapat mempunyai lebih dari satu label sekaligus. Berbeda dari klasifikasi multikelas, model multilabel dapat memberikan beberapa label untuk satu data.

### 1. Human Protein Atlas Image Classification

Kaggle: https://www.kaggle.com/competitions/human-protein-atlas-image-classification

- Tujuan: memprediksi lokasi protein di dalam sel manusia berdasarkan gambar mikroskopis.
- Fitur (X): gambar sel yang diambil menggunakan empat filter warna, yaitu merah, hijau, biru, dan kuning.
- Target (y): Target, yaitu kumpulan label lokasi protein pada setiap sampel.
- Label: terdapat 28 kemungkinan label, seperti Nucleoplasm, Nuclear membrane, Golgi apparatus, Mitochondria, dan Cytosol.
- Identifikasi: termasuk Multilabel Classification karena satu sampel dapat memiliki beberapa label lokasi protein sekaligus.
- Alasan dipilih: dataset ini secara langsung menerapkan konsep multilabel karena satu gambar dapat mewakili lebih dari satu lokasi protein.

### 2. FathomNet 2023

Kaggle: https://www.kaggle.com/competitions/fathomnet-out-of-sample-detection

- Tujuan: mengenali organisme laut dalam yang terdapat pada gambar bawah laut.
- Fitur (X): gambar bawah laut yang memperlihatkan satu atau beberapa organisme.
- Target (y): categories, yaitu kumpulan kategori organisme yang terdapat pada setiap gambar.
- Label: terdapat banyak kategori organisme laut, dengan anotasi hingga 290 kategori.
- Identifikasi: termasuk Multilabel Classification karena satu gambar dapat berisi beberapa organisme yang masing-masing mempunyai label berbeda.
- Alasan dipilih: dataset ini sesuai dengan konsep multilabel karena model perlu memprediksi seluruh kategori organisme yang muncul pada satu gambar, bukan hanya satu kategori.

### 3. Tradeshift Text Classification

Kaggle: https://www.kaggle.com/competitions/tradeshift-text-classification

- Tujuan: mengklasifikasikan dokumen berdasarkan beberapa kategori yang sesuai dengan karakteristik dokumen.
- Fitur (X): informasi teks dan dokumen yang telah direpresentasikan menjadi fitur numerik, kategorikal, atau fitur lain yang disediakan dalam dataset.
- Target (y): label dokumen yang tersedia pada file trainLabels.csv.
- Label: terdapat 33 label. Setiap label menunjukkan apakah suatu sampel termasuk dalam kategori tertentu, dengan nilai 1 berarti termasuk dan 0 berarti tidak termasuk.
- Identifikasi: termasuk Multilabel Classification karena satu dokumen dapat mempunyai beberapa label sekaligus. Jumlah label pada satu sampel tidak harus sama dengan satu.
- Alasan dipilih: dataset ini memiliki format target multilabel yang jelas sehingga cocok untuk mempelajari klasifikasi dokumen dengan lebih dari satu kategori.

## Kesimpulan

Berdasarkan sembilan dataset tersebut, Binary Classification digunakan untuk memprediksi dua kelas, seperti gagal bayar dan tidak gagal bayar. Multiclass Classification digunakan untuk memilih satu kelas dari beberapa kelas, seperti jenis tanaman, jenis makanan, atau jenis tutupan hutan. Sementara itu, Multilabel Classification digunakan ketika satu sampel dapat mempunyai beberapa label sekaligus, seperti gambar yang menunjukkan beberapa lokasi protein atau beberapa organisme laut.

Sebelum membangun model Machine Learning, fitur (X) dan target (y) perlu diperiksa terlebih dahulu. Untuk dataset gambar dan dokumen, fitur tidak selalu berupa kolom tabel biasa, melainkan dapat berupa gambar atau representasi teks. Pemilihan jenis klasifikasi harus mengikuti struktur label yang sebenarnya pada dataset.
